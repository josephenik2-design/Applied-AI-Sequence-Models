```python
import pandas as pd
import numpy as np

df = pd.read_csv("Variant_D_Fraud_Detection.csv")
display(df.head())
print("Shape:", df.shape)
```

```python
print(df.dtypes)
print("\nMissing:\n", df.isna().sum())
print("\nExact duplicates:", df.duplicated().sum())
display(df.describe())

for col in ["channel", "merchant_category", "country_risk_band", "device_changed_recently"]:
    print(f"\n{col}:\n", df[col].value_counts(dropna=False))
```

```python
d = df.drop_duplicates().copy()

# standardize categories
d["channel"] = d["channel"].str.strip().str.lower().map({"mobile":"Mobile","web":"Web","pos":"POS"})
d["merchant_category"] = d["merchant_category"].str.strip().str.title()
d["country_risk_band"] = d["country_risk_band"].str.strip().str.title()

# invalid values -> missing
d.loc[d["amount"] <= 0, "amount"] = np.nan
d.loc[~d["hour_utc"].between(0, 23), "hour_utc"] = np.nan
d.loc[~d["device_changed_recently"].isin([0, 1]), "device_changed_recently"] = np.nan

# missing label: drop (a label can't be invented)
d = d.dropna(subset=["fraud_label"])

# other missing values: median for numeric, mode for categorical
for c in ["amount", "hour_utc"]:
    d[c] = d[c].fillna(d[c].median())
for c in ["channel", "device_changed_recently"]:
    d[c] = d[c].fillna(d[c].mode()[0])

print(d.shape, d.isna().sum().sum())
```

```python
# 1. Class imbalance
print(d["fraud_label"].value_counts())
print("Fraud rate:", d["fraud_label"].mean())

# 2. A "model" that never flags fraud
print("Accuracy of always predicting 'not fraud':", (d["fraud_label"] == 0).mean())

# 3. Do the two label columns agree?
print(pd.crosstab(d["fraud_label"], d["chargeback_confirmed_after_30d"]))

# 4. Fraud by group (bias / proxy check)
for col in ["country_risk_band", "merchant_category", "channel", "event_month"]:
    print(f"\n{col}")
    print(d.groupby(col)[["fraud_label", "chargeback_confirmed_after_30d"]].agg(["sum", "count"]))
```
