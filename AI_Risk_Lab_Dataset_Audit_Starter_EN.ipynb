```python
import pandas as pd
import numpy as np

DATA_FILE = "Variant_D_Fraud_Detection.csv"
TARGET_COLUMN = "fraud_label"

df = pd.read_csv(DATA_FILE)
print("Rows and columns:", df.shape)
display(df.head(10))

audit = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "missing_count": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(2),
    "unique_values": df.nunique(dropna=True)
})
display(audit)
print("Exact duplicate rows:", int(df.duplicated().sum()))

display(df.describe(include="number").T)
for col in df.select_dtypes(include=["object", "category", "string"]).columns:
    print(f"\n--- {col} ---")
    display(df[col].value_counts(dropna=False).head(25))

display(df[TARGET_COLUMN].value_counts(dropna=False).to_frame("count"))

# extra evidence for the report
d = df.drop_duplicates()
print(pd.crosstab(d["fraud_label"].fillna(-1), d["chargeback_confirmed_after_30d"]))
for col in ["country_risk_band", "channel", "merchant_category", "event_month"]:
    print(f"\n{col}")
    print(d.groupby(col)[["fraud_label", "chargeback_confirmed_after_30d"]].agg(["sum", "count"]))
```
